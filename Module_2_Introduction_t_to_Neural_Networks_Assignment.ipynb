**1. Activation Functions:**

● Define the sigmoid function, a popular activation function used in neural
networks.


Answer :

The sigmoid function is an activation function used in neural networks to introduce non-linearity into the model. It maps any input value to a range between 0 and 1. The sigmoid function is mathematically represented as \(\sigma(x)=\frac{1}{1+e^{-x}}\). It is commonly used in binary classification problems where the output can be interpreted as a probability.

● Define the tanh function, which is another common activation function that
returns values between -1 and 1.

Answer :

The tanh (hyperbolic tangent) function is a commonly used activation function in neural networks. It maps the input values to a range between -1 and 1. The mathematical formula is \(\tanh(x)=\frac{e^x-e^{-x}}{e^x+e^{-x}}\). The tanh function is centered around zero, making it useful in neural networks where negative and positive outputs are important.


● Define the relu function, which returns positive values as they are and zeroes out negative values.

Answer :

The ReLU (Rectified Linear Unit) function is a commonly used activation function in neural networks. It returns the input value when the input is positive and returns zero when the input is negative. It is mathematically represented as \(ReLU(x)=\max(0,x)\). ReLU is widely used in hidden layers of neural networks because it is simple to calculate and helps the network learn complex patterns.


**2. Sample Data :**

● Use the following sample data for training:
x1=3, x2=2, yactual=17

```python
import numpy as np
```

```python
x = np.array([[3, 2]])
y = np.array([[17]])
```

```python
w = np.array([[0.5],
              [0.5]])
b = 0.0
```

```python
learning_rate = 0.01
```

```python
for epoch in range(1000):

    y_pred = np.dot(x, w) + b

    error = y_pred - y

    dw = np.dot(x.T, error)
    db = np.sum(error)

    w = w - learning_rate * dw
    b = b - learning_rate * db
```

```python
y_pred = np.dot(x, w) + b
```

```python
print("Learned weights:", w)
print("Learned bias:", b)
print("Predicted value:", y_pred)
print("Actual value:", y)
```

**5. Back Propagation:**

● Compute the gradients of the error with respect to the weights.

```python
dw1 = -2 * (y_actual - y_pred) * x1
dw2 = -2 * (y_actual - y_pred) * x2
```

```python
print("Gradient of w1 =", dw1)
print("Gradient of w2 =", dw2)
```

● Update the weights using the computed gradients and a learning rate.

```python
w1 = w1 - learning_rate * dw1
w2 = w2 - learning_rate * dw2
```

```python
print("Initial prediction:", y_pred)
print("Initial error:", error)
print("Updated w1:", w1)
print("Updated w2:", w2)
```

**6. Training:**

● The above steps (Feed Forward to Back Propagation) are performed
iteratively (20 times in this example) to refine the weights and minimize
the error.


```python
import numpy as np
```

```python
x1 = 3
x2 = 2
y_actual = 17
```

```python
w1 = np.random.uniform(1, 10)
w2 = np.random.uniform(1, 10)
```

```python
learning_rate = 0.01
```

```python
for i in range(20):

    y_pred = x1 * w1 + x2 * w2

    error = (y_actual - y_pred) * (y_actual - y_pred)

    dw1 = -2 * (y_actual - y_pred) * x1
    dw2 = -2 * (y_actual - y_pred) * x2

    w1 = w1 - learning_rate * dw1
    w2 = w2 - learning_rate * dw2

    print("Iteration:", i + 1)
    print("Predicted y:", y_pred)
    print("Error:", error)
    print("w1:", w1)
    print("w2:", w2)
    print("--------------------")
```

**7. Visualization:**

● Plot the progression of the error accross epochs.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x1 = 3
x2 = 2
y_actual = 17

w1 = np.random.uniform(1, 10)
w2 = np.random.uniform(1, 10)

learning_rate = 0.01
```

```python
errors = []

# Training for 20 epochs
for epoch in range(20):

    y_pred = x1 * w1 + x2 * w2

    error = (y_actual - y_pred) * (y_actual - y_pred)

    errors.append(error)

    dw1 = -2 * (y_actual - y_pred) * x1
    dw2 = -2 * (y_actual - y_pred) * x2

    w1 = w1 - learning_rate * dw1
    w2 = w2 - learning_rate * dw2
```

```python
print("Final w1:", w1)
print("Final w2:", w2)
print("Final prediction:", y_pred)
print("Final error:", error)
```

```python
plt.plot(range(1, 21), errors, marker='o')
plt.xlabel("Epoch")
plt.ylabel("Squared Error")
plt.title("Error Progression Across Epochs")
plt.grid(True)
plt.show()
```
